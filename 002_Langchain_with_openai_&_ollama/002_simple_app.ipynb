```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPEN_API_KEY"] = os.getenv("OPEN_API_KEY")

os.environ["LANGCHAIN_API_KEY"] = os.getenv("LANGCHAIN_API_KEY") # this is use for langsmith tracking
os.environ["LANGCHAIN_TRACING_V2"] = "true"
os.environ["LANGCHAIN_PROJECT"] = os.getenv("LANGCHAIN_PROJECT") 
```

### Data Ingestion : From the website ==> we need to scrape the data

```python
from langchain_community.document_loaders import WebBaseLoader
```

```python
loader=WebBaseLoader("https://docs.smith.langchain.com/tutorials/Administrators/manage_spend")
loader
```

```python
docs = loader.load()
docs
```

### Divide the Data into chunks 

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap= 200)
documents = text_splitter.split_documents(docs)
```

```python
documents
```

### Convert into Vector using vector embedding 

```python
from langchain_openai import OpenAIEmbeddings
embedding = OpenAIEmbeddings()
```

### Store into Vector DB

```python
from langchain_community.vectorstores import FAISS
vectorstoredb = FAISS.from_documents(documents, embedding)
```
